# Data Exploration

## Overview

Data exploration is the first analytical stage of the Movie Analytics SQL Project. The purpose of this stage is to develop a clear understanding of the database structure, the data contained within each table, and the relationships that connect the different entities.

The database contains information covering:

- Movies
- Actors
- Cast members
- Genres
- Keywords
- Languages
- Production companies
- Production countries
- Oscar awards

Before performing transformations or building analytical models, it is important to understand the characteristics and structure of these datasets.

## Objectives

The data exploration process will focus on:

- Identifying all tables available in the database.
- Understanding the structure and purpose of each table.
- Examining columns and data types.
- Identifying primary and foreign key relationships.
- Understanding the grain of each table.
- Measuring the number of records within each table.
- Examining unique and categorical values.
- Identifying potential missing and duplicate records.
- Understanding one-to-one, one-to-many, and many-to-many relationships.
- Identifying potential issues that could affect downstream analysis.

## Analytical Approach

The exploration will be performed using **SQL through SQLAlchemy**, with **Pandas** used to retrieve and inspect query results within the Jupyter Notebook.

The analysis will begin at the database level and progressively move toward table-level and column-level exploration.

Particular attention will be given to bridge tables because they represent **many-to-many relationships** and can introduce row multiplication when multiple tables are joined.

## Exploration Process

The exploration will cover the following areas:

1. Database and table inventory
2. Table structure and schema
3. Primary and foreign key relationships
4. Record counts
5. Column-level characteristics
6. Unique values and cardinality
7. Missing values
8. Duplicate records
9. Numerical and categorical fields
10. Relationships between tables

## Expected Outcome

At the end of the exploration stage, there should be a clear understanding of:

- What data is available.
- How the tables are structured.
- How the entities relate to one another.
- The grain of each table.
- Which fields can be used as keys.
- Potential data quality issues.
- Potential data modeling considerations.

The findings from this stage will provide the foundation for the subsequent stages of the project:

1. **Data Profiling**
2. **Data Quality Assessment**
3. **Data Modeling**
4. **Analytics Mart Development**
5. **Exploratory Data Analysis**
6. **Business Analysis and Insights**

> **Key Principle:** Before transforming or analyzing data, understand its structure, grain, relationships, and quality.

```python
# Importing the required libraries
import pandas as pd
import matplotlib.pyplot as plt
from  sqlalchemy import create_engine
from pathlib import Path

print("All libraries imported successfully!")
```

```python
# Creating a database engine
db_path = Path.home() / "Documents" / "Movie DB SQL Analytics" / "Database" / "TMDB-a-4006.db"

engine = create_engine(f"sqlite:///{db_path}")
```

```python
# Testing the database connection
query = """
SELECT sqlite_version();
"""

pd.read_sql(query, engine)
```

```python
# Checking the tables in the database
query = """
SELECT name
FROM sqlite_master
WHERE type='table'
ORDER BY name;
"""

tables = pd.read_sql(query, engine)

tables
```

```python
# Number of tables in the database
len(tables)
```

```python
# Counting the number of records in each table
record_counts = {}
for table in tables['name']:
    query = f"SELECT COUNT(*) AS count FROM {table};"
    count = pd.read_sql(query, engine).iloc[0]['count']
    record_counts[table] = count
    print(f"Number of records in {table}: {count}")
```
