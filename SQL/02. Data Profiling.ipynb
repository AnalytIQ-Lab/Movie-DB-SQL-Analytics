# Data Profiling

## Overview

Data profiling is the process of systematically examining the structure, completeness, consistency, and characteristics of the data before performing transformation or analysis.

The objective of this stage is to establish a clear understanding of the quality and distribution of the data across the movie database.

## Objectives

The data profiling process will focus on:

- Measuring the number of records in each table.
- Identifying the number of columns in each table.
- Examining column names and data types.
- Measuring the number of unique values.
- Identifying missing or null values.
- Detecting potential duplicate records.
- Examining numerical data distributions.
- Reviewing categorical value distributions.
- Identifying minimum and maximum values where applicable.
- Understanding the cardinality of key fields.
- Identifying potential data anomalies that require further investigation.

## Profiling Dimensions

The database will be profiled across the following dimensions:

### 1. Completeness

Determine whether required fields contain missing or null values and measure the percentage of incomplete records.

### 2. Uniqueness

Determine whether primary key fields contain duplicate values and assess the uniqueness of important business attributes.

### 3. Validity

Examine whether values conform to their expected data types, formats, and business meaning.

### 4. Consistency

Identify inconsistent values, naming conventions, or categorical representations across the database.

### 5. Distribution

Analyze the distribution of numerical and categorical fields to understand the characteristics of the underlying data.

### 6. Cardinality

Measure the number of distinct values within important fields to understand their uniqueness and relationship to the table grain.

## Analytical Approach

The profiling will be performed using **SQL through SQLAlchemy**, with **Pandas** used to retrieve and inspect profiling results within the Jupyter Notebook.

SQL queries will be used to calculate record counts, distinct values, null values, duplicates, and statistical measures. Pandas will then be used to organize and review the results.

## Expected Outcome

The profiling exercise will establish a baseline understanding of the database and highlight areas requiring further investigation during the **Data Quality Assessment** stage.

The results will help determine:

- Which tables require additional validation.
- Which columns contain missing values.
- Whether key fields are truly unique.
- Where potential anomalies exist.
- Which fields require transformation or standardization.
- How the data should be handled when developing the analytics marts.

> **Key Principle:** Data profiling establishes an evidence-based understanding of the data before any cleaning, transformation, or analytical modeling is performed.

```python
# Importing the required libraries
import pandas as pd
from  sqlalchemy import create_engine
from pathlib import Path

print("All libraries imported successfully!")
```

```python
# Creating a database engine
db_path = Path.home() / "Documents" / "Movie DB SQL Analytics" / "Database" / "TMDB-a-4006.db"

engine = create_engine(f"sqlite:///{db_path}")

print("Database Connection established successfully!")
```

# Profiling the Actor Table
### - Row Count
### - Distinct count
### - Missing Value
### - Duplicate Values

```python
# Row count of the actors table
pd.read_sql("""
SELECT COUNT(*) AS Total
FROM actors;
""", engine)
```

```python
# Distinct count of actors in the actors table
pd.read_sql("""
SELECT COUNT(DISTINCT actor_id)
FROM actors;
""", engine)
```

```python
# Missing values in the actors table
pd.read_sql("""
SELECT COUNT(*)
FROM actors
WHERE actor_name IS NULL;
""", engine)
```

```python
# Duplicate actor_id values in the actors table
pd.read_sql("""
SELECT actor_id,
COUNT(*) AS Total
FROM actors
GROUP BY actor_id
HAVING COUNT(*) > 1;
""", engine)
```
